```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python

plt.figure(figsize=(7.5,3), dpi=250)
plt.subplot(121)
df_steven = pd.read_csv("../../outputs/trapezoid-2025-10-26-20-42-26-steven-step42/records.csv")
df_normal = pd.read_csv("../../outputs/trapezoid-2025-10-26-20-58-15-nu049-step42/records.csv")
steven_bottom_x = df_steven["bottom_x"]
steven_bottom_y = df_steven["bottom_y"]
normal_bottom_x = df_normal["bottom_x"]
normal_bottom_y = df_normal["bottom_y"]

plt.plot(0.5/50*(np.arange(len(steven_bottom_x)+1)), [0.0]+list(steven_bottom_x), c="#5171A5", lw=2.5, label="Steven's function (x-dir)")
plt.plot(0.5/50*(np.arange(len(steven_bottom_y)+1)), [0.0]+list(steven_bottom_y), c="#92374D", lw=2.5, label="Steven's function (y-dir)")
plt.plot(0.5/50*(np.arange(len(normal_bottom_x)+1)), [0.0]+list(normal_bottom_x), c="#5171A5", ls=":", lw=3.5, label="neo-Hookean (x-dir)")
plt.plot(0.5/50*(np.arange(len(normal_bottom_y)+1)), [0.0]+list(normal_bottom_y), c="#92374D", ls=":", lw=3.5, label="neo-Hookean (y-dir)")

plt.xlabel("Displacement (mm)")
plt.ylabel("Traction ($N$)")
plt.title(r"$c=1$")
plt.legend(frameon=False)

plt.subplot(122)

df_steven = pd.read_csv("../../outputs/trapezoid-2025-10-26-21-46-49-steven-c0-step46/records.csv")
df_normal = pd.read_csv("../../outputs/trapezoid-2025-10-26-21-46-06-nu049-c0-step43/records.csv")
steven_bottom_x = df_steven["bottom_x"]
steven_bottom_y = df_steven["bottom_y"]
normal_bottom_x = df_normal["bottom_x"]
normal_bottom_y = df_normal["bottom_y"]

plt.plot(0.5/50*(np.arange(len(steven_bottom_x)+1)), [0.0]+list(steven_bottom_x), c="#5171A5", lw=2.5, label="Steven's function (x-dir)")
plt.plot(0.5/50*(np.arange(len(steven_bottom_y)+1)), [0.0]+list(steven_bottom_y), c="#92374D", lw=2.5, label="Steven's function (y-dir)")
plt.plot(0.5/50*(np.arange(len(normal_bottom_x)+1)), [0.0]+list(normal_bottom_x), c="#5171A5", ls=":", lw=3.5, label="neo-Hookean (x-dir)")
plt.plot(0.5/50*(np.arange(len(normal_bottom_y)+1)), [0.0]+list(normal_bottom_y), c="#92374D", ls=":", lw=3.5, label="neo-Hookean (y-dir)")

plt.xlabel("Displacement (mm)")
plt.ylabel("Traction ($N$)")
plt.legend(frameon=False)
plt.title(r"$c=0$")
plt.tight_layout()
plt.savefig("trapezoid_reaction_displacement_compare.pdf")
plt.show()
plt.close()
```
