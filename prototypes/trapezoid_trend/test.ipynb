```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(4,3.5), dpi=250)
ax = plt.subplot(111)
df = pd.read_csv("../../outputs/trapezoid-2025-10-29-10-25-31/records_opt.csv")
ax.plot(np.arange(len(df)), -df["top_x"]*5e-5, ls="-", c="#92374D", label="Leftward force")
ax.plot(np.arange(len(df)), -df["top_y"]*5e-5, ls="-", c="#5171A5", label="Downward force")

ax.legend(frameon=False)
handles, labels = ax.get_legend_handles_labels()
ax.get_legend().remove()

ax.set_xlim(left=0.0)
ax.set_ylim([0.05, 0.7])
ax.set_xlabel("Iteration")
ax.set_ylabel(r"Loading on the imaginary plate ($N$)")

ax = ax.twinx()
ax.plot(np.arange(len(df)), -(df["top_y"]-df["top_x"])*5e-5, c="gray", ls="--", label="Difference (Loss)")
ax.set_ylabel("Difference between\ndownward and leftward forces" +r" ($N$)")
ax.set_ylim([0.3, 0.5])
ax.set_yticks([0.3, 0.35, 0.4, 0.45, 0.5])
ax.legend(frameon=False)
handles2, labels2 = ax.get_legend_handles_labels()
ax.get_legend().remove()
ax.legend(handles=handles+handles2, labels=labels+labels2, frameon=False)
plt.show()
plt.close()
```
