```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(4,3), dpi=250)
plt.subplot(111)
df = pd.read_csv("../../outputs/trapezoid-2025-10-29-10-25-31/records_opt.csv")
plt.plot(np.arange(len(df)), -df["top_x"]*5e-5, ls="-", c="#92374D", label="Leftward force")
plt.plot(np.arange(len(df)), -df["top_y"]*5e-5, ls="-", c="#5171A5", label="Downward force")

plt.legend(frameon=False)
plt.xlabel("Iteration")
plt.ylabel(r"Loading on the imaginary plate ($N$)")
plt.show()
plt.close()
```
