```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.DataFrame({"Case": ["Topology\n(without $I_1$ constraint)", "Composition ratio", "Topology", "Topology + Composition ratio"], "Strain energy": [0.00475187, 0.003537, 0.0041394, 0.0042632]})

plt.figure(figsize=(5.,3.2), dpi=250)
ax = plt.subplot(111)
sns.barplot(data=df, x="Case", y="Strain energy", palette=["#E8E8E8", "#1768AC", "#A76571", "#F4A261"], ax=ax)
# ax.set_ylim([0, 6])
# handles, labels = ax.get_legend_handles_labels()
# ax.get_legend().remove()
# ax.legend(handles=handles, labels=labels, title=False, frameon=False, ncol=1)
ax.set_xticklabels(ax.get_xticklabels(), rotation=-20, ha="left")
plt.text(ax.get_xticks()[0], ax.get_ylim()[1]/2, "Baseline", rotation=90, ha="center", va="center", fontsize=16, color="#C3C3C3", fontweight="bold")
ax.set_xlabel(None)
ax.set_ylabel("Strain energy\nafter optimization (mJ)")

plt.tight_layout()
plt.savefig("clamp_compare_energy.pdf", bbox_inches="tight")
plt.show()
plt.close()
```
